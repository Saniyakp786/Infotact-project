```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/Cleaned_Dataset_IntermodalRail.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
print("Vibration Level Information:")
print(df["vibration_level"].describe())
```

```python
vibration_by_spoilage = df.groupby(
    "spoilage_status"
)["vibration_level"].mean()

print("Average Vibration Level by Spoilage Status:")
print(vibration_by_spoilage)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["vibration_level"],
    df["spoilage_risk_percent"]
)

plt.title("Vibration vs Spoilage Risk")
plt.xlabel("Vibration Level")
plt.ylabel("Spoilage Risk (%)")

plt.tight_layout()
plt.show()
```

```python

vibration_risk_correlation = df[
    "vibration_level"
].corr(
    df["spoilage_risk_percent"]
)

print(
    "Vibration vs Spoilage Risk Correlation:",
    vibration_risk_correlation
)
```

```python

plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="vibration_level",
    y="spoilage_risk_percent"
)

plt.title("Vibration vs Spoilage Risk")
plt.xlabel("Vibration Level")
plt.ylabel("Spoilage Risk (%)")

plt.tight_layout()
plt.show()
```
