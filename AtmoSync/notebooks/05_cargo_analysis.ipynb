```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/Cleaned_Dataset_IntermodalRail.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df["cargo_type"].value_counts()
```

```python
cargo_risk = df.groupby(
    "cargo_type"
)["spoilage_risk_percent"].mean().sort_values(
    ascending=False
)

cargo_risk
```

```python
cargo_risk.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Average Spoilage Risk by Cargo Type")
plt.xlabel("Cargo Type")
plt.ylabel("Spoilage Risk (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
