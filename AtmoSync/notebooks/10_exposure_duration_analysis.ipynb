```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/Cleaned_Dataset_IntermodalRail.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
print("Exposure Duration Information:")
print(df["exposure_duration_hours"].describe())
```

```python
exposure_by_spoilage = df.groupby(
    "spoilage_status"
)["exposure_duration_hours"].mean()

print("Average Exposure Duration by Spoilage Status:")
print(exposure_by_spoilage)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["exposure_duration_hours"],
    df["spoilage_risk_percent"]
)

plt.title("Exposure Duration vs Spoilage Risk")
plt.xlabel("Exposure Duration (Hours)")
plt.ylabel("Spoilage Risk (%)")

plt.tight_layout()
plt.show()
```

```python
exposure_risk_correlation = df[
    "exposure_duration_hours"
].corr(
    df["spoilage_risk_percent"]
)

print(
    "Exposure Duration vs Spoilage Risk Correlation:",
    exposure_risk_correlation
)
```
