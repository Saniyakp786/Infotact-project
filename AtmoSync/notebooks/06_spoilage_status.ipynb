```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/Cleaned_Dataset_IntermodalRail.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
status_count = df["spoilage_status"].value_counts()

status_count
```

```python
status_count.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Shipment Spoilage Status")
plt.xlabel("Status")
plt.ylabel("Number of Shipments")
plt.tight_layout()
plt.show()
```
