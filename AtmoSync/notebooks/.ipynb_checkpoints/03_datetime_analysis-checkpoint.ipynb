```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/Cleaned_Dataset_IntermodalRail.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df["recorded_datetime"] = pd.to_datetime(
    df["recorded_datetime"],
    errors="coerce"
)

df["recorded_date"] = pd.to_datetime(
    df["recorded_date"],
    errors="coerce"
)
```

```python
df["year"] = df["recorded_datetime"].dt.year
df["month"] = df["recorded_datetime"].dt.month
df["day"] = df["recorded_datetime"].dt.day
df["hour"] = df["recorded_datetime"].dt.hour
```
