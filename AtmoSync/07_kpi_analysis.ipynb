```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/cleaned/Cleaned_Dataset_AirFreight.csv")

df.head()
```

```python
df.columns.tolist()
```

```python
kpis = {
    "Total Shipments": len(df),
    "Average Temperature": df["temperature_celsius"].mean(),
    "Average Humidity": df["humidity_percentage"].mean(),
    "Average Spoilage Risk": df["spoilage_risk_percent"].mean(),
    "At Risk Shipments": (
        df["spoilage_risk_percent"] > 60
    ).sum(),
    "Average Shelf Life": (
        df["remaining_shelf_life_days"].mean()
    )
}

for key, value in kpis.items():
    print(f"{key}: {value:.2f}")
```

```python
print(f"Total Shipments: {kpis['Total Shipments']}")
print(f"Average Temperature: {kpis['Average Temperature']:.2f}")
print(f"Average Humidity: {kpis['Average Humidity']:.2f}")
print(f"Average Spoilage Risk: {kpis['Average Spoilage Risk']:.2f}")
print(f"At Risk Shipments: {kpis['At Risk Shipments']}")
print(f"Average Shelf Life: {kpis['Average Shelf Life']:.2f}")
```

```python
kpi_table = pd.DataFrame({
    "KPI": [
        "Total Shipments",
        "Average Temperature",
        "Average Humidity",
        "Average Spoilage Risk",
        "At Risk Shipments",
        "Average Shelf Life"
    ],
    "Value": [
        kpis["Total Shipments"],
        kpis["Average Temperature"],
        kpis["Average Humidity"],
        kpis["Average Spoilage Risk"],
        kpis["At Risk Shipments"],
        kpis["Average Shelf Life"]
    ]
})

kpi_table
```
