```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/cleaned/Cleaned_Dataset_AirFreight.csv")

print(df.shape)
print(df.head())
```

```python
print(df["temperature_celsius"].describe())
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["temperature_celsius"], bins=20)

plt.xlabel("Temperature (°C)")
plt.ylabel("Number of Records")
plt.title("Temperature Distribution")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    df["temperature_celsius"],
    df["spoilage_risk_percent"]
)

plt.xlabel("Temperature (°C)")
plt.ylabel("Spoilage Risk (%)")
plt.title("Temperature vs Spoilage Risk")

plt.show()
```

```python
correlation = df["temperature_celsius"].corr(
    df["spoilage_risk_percent"]
)

print("Temperature vs Spoilage Risk Correlation:", correlation)
```
