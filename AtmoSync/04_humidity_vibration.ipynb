```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("data/cleaned/Cleaned_Dataset_AirFreight.csv")

# Humidity vs Spoilage Risk
plt.figure(figsize=(10, 5))

plt.scatter(
    df["humidity_percentage"],
    df["spoilage_risk_percent"]
)

plt.xlabel("Humidity (%)")
plt.ylabel("Spoilage Risk (%)")
plt.title("Humidity vs Spoilage Risk")

plt.show()
```

```python
fig = plt.figure(figsize=(10, 5))

plt.scatter(
    df["vibration_level"],
    df["spoilage_risk_percent"]
)

plt.xlabel("Vibration")
plt.ylabel("Spoilage Risk (%)")
plt.title("Vibration vs Spoilage Risk")

plt.show()
```
