```python
import pandas as pd

print("Pandas imported successfully!")
```

```python
df = pd.read_csv("data/cleaned/Cleaned_Dataset_AirFreight.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
print("Data types:")
print(df.dtypes)
```

```python
print("Column names:")
for col in df.columns:
    print(col)
```

```python
numeric_columns = [
    "temperature_celsius",
    "humidity_percentage",
    "vibration_level",
    "pressure_kpa",
    "exposure_duration_hours",
    "door_open_count",
    "salinity_ppt",
    "thermal_leakage_score",
    "spoilage_risk_percent",
    "remaining_shelf_life_days"
]

for col in numeric_columns:
    if col in df.columns:
        print(
            col,
            "min =", df[col].min(),
            "max =", df[col].max()
        )
```

```python
df.describe()
```

```python
print("Negative values check:")

for col in numeric_columns:
    if col in df.columns:
        negative_count = (df[col] < 0).sum()
        print(col, "=", negative_count)
```

```python
print("===================================")
print("      DATA QUALITY SUMMARY")
print("===================================")

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nTotal duplicate rows:")
print(df.duplicated().sum())

print("\nNumeric columns checked:")
for col in numeric_columns:
    if col in df.columns:
        print("-", col)

print("\nData quality check completed successfully!")
```
